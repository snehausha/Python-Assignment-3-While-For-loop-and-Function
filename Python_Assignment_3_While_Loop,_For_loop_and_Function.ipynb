## TASK 1: Number Guessing Game

While loop & Control Statements :
(else, break, continue, pass)

Created a Python program that implements a simple number
guessing game using a while loop. The program made use of control
statements such as else, break, and continue.


###STEPS:

Used **Import random and randint function** to generate a random number between 1 and 10 that the user has to guess.

created a variable **attempts** to 3, which represents the maximum number of guesses allowed.

To implement the Guessing Logic, used **while loop** to keep guessing until they get the correct number or run out of attempts.

Provided feedback to the user for each guess:

➢ If the guess is out of the valid range (1 to 10), a feedback of "INVALID RANGE, PLEASE TRY AGAIN!"  will shows up.

➢ If the guess is greater than the secret number, "Oops!You've guessed too high." will shows up.

➢ If the guess is lower than the secret number,"Oops!You've guessed too low." will shows up.

➢ If the guess is correct, it will congratulate the user and end the game.

After 3 tries,(excluding the input of invalid range) it will show a message of "GAME OVER","Better luck next time!!!"

**Control Statements** such as;

**continue** is used to skip the rest of the loop after informing the user if the guess is out of range.

**break** used to exit the loop when the guess is correct.

**else** with the while loop to provide a message like "Better luck next time!" if the user runs out of attempts without guessing the correct number.
















```python
import random
secret_number=random.randint(1,10)
attempt=0
while attempt<3:
    guess=int(input("Guess a number from 1-10: "))
    if guess<1 or guess>10:
      print("INVALID RANGE,PLEASE TRY AGAIN!")
      continue
    attempt+=1

    if guess==secret_number:
        print("Congratulations,You guessed it right!!")
        break
    elif guess > secret_number:
        print("Oops!You've guessed too high.")
    else:
        print("Oops!You've guessed too low.")

    if attempt==3:
     print("Game Over")


else:
  print("Better luck next time!!!")
```

##Task 2: Multiplication Table Generator

Created a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range function.

The **input()** function is used to accept the **number (int())**,  **for loop** repeats the multiplication operation through the **range** given (1,11) and **print()** displays the output as given.

```python
num=int(input("Enter a number for which you want multiplication table: "))
for i in range(1,11):
  print(num,"x", i,"=", num*i)

```

##TASK 3: BMI CALCULATOR

Created a Python program that calculates the Body Mass Index
(BMI).

The **input()** function used to enter and store the numbers **(float())** in two parameters weight and height respectively, which was created under the function named **calculate_bmi()**.
The calculated bmi (weight/height**2) is returned from the function using **return bmi**.
**print()** displays the calculated output, rounded to two decimal points using **round(_,2)**



```python
weight=float(input("Enter your weight in kgs: "))
height=float(input("Enter your height in meters: "))
def calculate_bmi(weight,height):
  bmi=(weight)/(height**2)
  return bmi
print("Your BMI is:",round(weight/(height**2),2))

```
